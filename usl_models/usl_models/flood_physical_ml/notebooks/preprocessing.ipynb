```python
for city in city_dic_list:
    
    city_name = city["city_name"]
    max_value = city["max"]
    min_value = city["min"]

    city_dir_root = root_dir / city_name

    files = city_dir_root.glob("**/geospatial/**/*")
    for f in files:
        print(f)
        file = np.load(f)
        dem = file[:, :, 0]
        new_dem = scale_back_dem(dem, max_value, min_value)
        file[:, :, 0] = new_dem
        np.save(f, file)
```

```python
for city in city_dic_list:
    
    city_name = city["city_name"]
    max_value = city["max"]
    min_value = city["min"]

    city_dir_root = root_dir / city_name

    files = city_dir_root.glob("**/geospatial/**/*")
    for f in files:
        print(f)
        file = np.load(f)
        dem = file[:, :, 0]
        dem[dem<min_value-10] = min_value-10
        file[:, :, 0] = dem
        np.save(f, file)
```
